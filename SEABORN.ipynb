#**SEABORN**

```python
import seaborn as sns
import matplotlib.pyplot as plt
data={
    "Product":["Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
    "Sales":[50,80,40,70,90,60],
    "Price":[50000,3000,20000,55000,32000,22000],
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar"]
}

#BARPLOT
sns.barplot(x="Product",y="Sales",data=data)                                    #Compare the sales of each product
plt.show()
print("\n")

#COUNTPLOT
sns.countplot(x="Product",data=data)                                            #Count how many times each product appears
plt.show()
print("\n")

#HISTPLOT
sns.histplot(x="Product",y="Sales",data=data)
plt.show()
print("\n")

#BOXPLOT
sns.boxplot(y=data["Sales"])                                                    #Show the spread of sales and possible outliers
plt.show()
print("\n")

#SCATTERPLOT
sns.scatterplot(x="Price",y="Sales",data=data)                                  #Show the relationship between price and sales
plt.show()
print("\n")

#linePLOT
sns.lineplot(x="Month",y="Sales",data=data)                                     #Show the sales trend by month
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data=np.random.randn(1000)
plt.figure(figsize=(8,6))
sns.histplot(data,bins=30,kde=True,color="blue",edgecolor="red")                #kde=Kernal Density Estimation(Curve of graph)  &  bins=Number of partitions in the graph
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Distribution of random data using seaborn")
plt.grid(axis="y",alpha=0.75)
plt.show()                                      #markers =  . , * , o , s , p , ^    &   linestyle = -. , : , --
```
