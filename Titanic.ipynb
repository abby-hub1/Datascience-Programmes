```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/titanic.csv")                                                  #To read CSV file.
df                                                                              #Print the file
```

```python
print(df.shape)                                                                 #Show number of rows and columns
```

```python
df.describe()                                                                   #Gives the ststistical summaryof numerical columns,includeing Mean,Minimum,Maximum and Quartiles
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(                                           #Replace missing values
    df["Embarked"].mode()[0]                                                    #Find the most frequent value
)
```

```python
print(df.isnull().sum())
print("Before : ",df.shape[0])
df=df.drop_duplicates()
print("After : ",df.shape[0])
```

```python
df=df.rename(columns={
    "Passengerid" : "PassengerID",
    "2urvived" : "Survived"
})
```

```python
df["Sex"]=df["Sex"].map({
    0 : "Male",
    1 : "Female"
})
df.head()
```

```python
sns.histplot(data=df,x="Age",bins=20)
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
sns.countplot(data=df,x="Sex")
plt.title("Titanic Survivor by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.show()
sns.countplot(data=df,x="Pclass")
plt.title("Titanic Survival by passenger class")
plt.xlabel("Passenger class")
plt.ylabel("Number of passengers")
plt.show()
```
