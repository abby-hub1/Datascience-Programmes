```python
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([5,6,7,8])
result=arr1+arr2
print(result)
print(arr1.ndim)
print(arr2.ndim)
```

```python
#Conversion of one dimentional array to two dimensional array b   -   "reshape" is used
import numpy as np
array1=np.array([1,2,3,4,5,6,7,8])
array2=array1.reshape(2,4)
print(array1)
print(array2)
print(array2.shape)
```

PANDAS - Panel Data System


PANDAS is an a open source python library.
* Used for data cleaning, data analysis, data manupulation and data visualisation prepration.
* Working with EXCEL sheet or tabular formatted data with rows and columns.

PANDAS DATA STRUCTURE
* PANDAS are mainly two types - Series and data frames
* Import - IMPORT PANDAS as pd



SERIES
* Series is an one dimensional labeled array

```python
import pandas as pd
marks=pd.Series([1,2,3,4,5])
print(marks)
```

```python
import pandas as pd
marks=pd.Series([10,20,30,40],index=["Maths","CS","AI","SCIENCE"])
print(marks)
```

```python
import pandas as pd
age=pd.Series([18,17,19],index=["Akhil","Arun","Abhi"])
print(age)
```

```python
import pandas as pd
depart=pd.Series(["BSC","BCA","BBA"],index=["Abhi","Akhil","Arun"])
print(depart)
```

```python
import pandas as pd
depart=pd.Series(["BSC","BCA","BBA"],index=["Abhi","Akhil","Arun"])
print(depart)
print(depart["Abhi"])
```
